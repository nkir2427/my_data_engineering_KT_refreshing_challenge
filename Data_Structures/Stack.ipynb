1.Write a function in python that can reverse a string using stack data structure. Use Stack class from the tutoria

    reverse_string("We will conquere COVID-19") should return "91-DIVOC ereuqnoc lliw eW"


```python
from collections import deque
stack = deque()
```

```python
class Stack:
    def __init__(self):
        self.container = deque()
    
    def push(self,val):
        self.container.append(val)
        
    def pop(self):
        return self.container.pop()
    
    def peek(self):
        return  self.container[-1]
    
    def is_empty(self):
        return len(self.container)==0
    
    def size(self):
        return len(self.container)

    def reverse_string(self,text):
        for ch in text:
            self.container.append(ch)

        reverse=""

        while not self.is_empty():
            reverse += self.pop()

        return reverse
```

2.Write a function in python that checks if paranthesis in the string are balanced or not. Possible parantheses are "{}',"()" or "[]". Use Stack class from the tutorial.

        is_balanced("({a+b})")     --> True
        is_balanced("))((a+b}{")   --> False
        is_balanced("((a+b))")     --> True
        is_balanced("))")          --> False
        is_balanced("[a+b]*(x+2y)*{gg+kk}") --> True

```python
class Stack1:
    def __init__(self):
        self.container = deque()
    
    def push(self,val):
        self.container.append(val)
        
    def pop(self):
        return self.container.pop()
    
    def peek(self):
        return  self.container[-1]
    
    def is_empty(self):
        return len(self.container)==0
    
    def size(self):
        return len(self.container)

    def is_balanced1(self,expression):
        stack = Stack()
    
        matching_brackets = {
            ')': '(',
            '}': '{',
            ']': '['
        }
    
        for character in expression:
    
            # Push opening brackets
            if character in "({[":
                stack.push(character)
    
            # Check closing brackets
            elif character in ")}]":
    
                # No opening bracket is available
                if stack.is_empty():
                    return False
    
                opening_bracket = stack.pop()
    
                # Opening and closing brackets do not match
                if opening_bracket != matching_brackets[character]:
                    return False
    
        # Balanced only when no opening brackets remain
        return stack.is_empty()
     

    def is_balanced(self,val):

        for ch in val:
            self.container.append(ch)

        round_count=0
        curly_count=0
        square_count=0
        round_count1=0
        curly_count1=0
        square_count1=0

        for ch in self.container:
            if ch in '(':
                round_count += 1
            elif ch in '{':
                curly_count += 1
            elif ch in '[':
                square_count += 1
            elif ch in ')':
                round_count1 += 1
            elif ch in '}':
                curly_count1 += 1
            elif ch in ']':
                square_count1 += 1   
        
        return round_count==round_count1 and  square_count== square_count1  and  curly_count==curly_count1
```
