4. Implement hash table where collisions are handled using linear probing. We learnt about linear probing in the video tutorial. Take the hash table implementation that uses chaining and modify methods to use linear probing. Keep MAX size of arr in hashtable as 10

````python
class HashTable:  
    def __init__(self):
        self.MAX = 10
        self.arr = [None for i in range(self.MAX)]
        
    def get_hash(self, key):
        hash = 0
        for char in key:
            hash += ord(char)
        return hash % self.MAX
    
    def __getitem__(self, key):
        arr_index = self.get_hash(key)
        
        while self.arr[arr_index] is not None:

            if self.arr[arr_index][0] == key:
                return self.arr[arr_index][1]
            arr_index = (arr_index + 1) % self.MAX
            
            
    def __setitem__(self, key, val):
        h = self.get_hash(key)
    
        while self.arr[h] is not None and self.arr[h][0] != key:
            h = (h + 1) % self.MAX
    
        self.arr[h] = (key, val)
        


    def __delitem__(self, key):
        arr_index = self.get_hash(key)
    
        print("Hash index:", arr_index)
        print("Before:", self.arr[arr_index])

        while self.arr[arr_index] is not None:
    
            if self.arr[arr_index][0] == key:
                self.arr[arr_index] = None
            arr_index = (arr_index + 1) % self.MAX
    
        print("After:", self.arr[arr_index])

        
````

## Linear Probing for Collision Handling

A **collision** occurs in a hash table when two or more keys are mapped to the same hash index. One common technique for resolving collisions is **linear probing**, which belongs to the **open addressing** family of collision resolution methods.

In linear probing, each bucket stores **only one key-value pair**. If the bucket computed by the hash function is already occupied, the algorithm searches the next available bucket sequentially until an empty location is found. The probing sequence is:

```
(hash_index + 0) % table_size
(hash_index + 1) % table_size
(hash_index + 2) % table_size
...
```

During **search** and **deletion**, the algorithm starts from the original hash index and follows the same probing sequence until the desired key is found or an empty bucket is encountered.

### Example

Suppose the hash function returns index **4** for both `"apple"` and `"banana"`.

| Index | Bucket |
|------:|:-------|
| 4 | ("apple", 10) |
| 5 | Empty |

When inserting `"banana"`:

1. Compute `hash("banana") = 4`.
2. Bucket 4 is occupied.
3. Check bucket 5.
4. Bucket 5 is empty, so insert `("banana", 20)` there.

The original hash value of `"banana"` remains **4**; only its storage location changes due to linear probing.
