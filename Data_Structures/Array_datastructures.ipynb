### Exercise: Array DataStructure

Let us say your expense for every month are listed below,

| month    | expenses |
|----------|-----------:|
| January  | 2200       |
| February | 2350       |
| March    | 2600       |
| April    | 2130       |
| May      | 2190       |


Create a list to store these monthly expenses and using that find out,

1. In Feb, how many dollars you spent extra compare to January?
2. Find out your total expense in first quarter (first three months) of the year.
3. Find out if you spent exactly 2000 dollars in any month
4. June month just finished and your expense is 1980 dollar. Add this item to our monthly expense list
5. You returned an item that you bought in a month of April and
got a refund of 200$. Make a correction to your monthly expense list
based on this

```python
monthly_expenses=[2200,2350,2600,2130,2190]
print(monthly_expenses)
```

```python
#1. In Feb, how many dollars you spent extra compare to January?

extra_spent_feb=monthly_expenses[1]-monthly_expenses[0]
print(extra_spent_feb)
```

```python
#2. Find out your total expense in first quarter (first three months) of the year.

i=0
total=0
for i in range(len(monthly_expenses)):
    if i==3:
        break
    else:
         total +=monthly_expenses[i]

print(total)
```

```python
#3. Find out if you spent exactly 2000 dollars in any month

dolors_any_month=(2000 in monthly_expenses)
print(dolors_any_month)
```

```python
#4. June month just finished and your expense is 1980 dollar. Add this item to our monthly expense list

monthly_expenses.append(1980)
print(monthly_expenses)
```

```python
#5. You returned an item that you bought in a month of April and got a refund of 200$. Make a correction to your monthly expense list based on this
print("before refund:", monthly_expenses)
monthly_expenses[3]+=200
print("after refund:", monthly_expenses)
```


### 2.You have a list of your favourite marvel super heros.

    heros=['spider man','thor','hulk','iron man','captain america']

Using this find out,

1. Length of the list

   so remove it from the list first and then add it after 'hulk'

```python
heros=['spider man','thor','hulk','iron man','captain america']
print(heros)
```

```python
#1. Length of the list

print("length of the list:", len(heros))
```

```python
#2. Add 'black panther' at the end of this list

heros.append('black panther')
print(heros)
```

```python
#3. You realize that you need to add 'black panther' after 'hulk',   so remove it from the list first and then add it after 'hulk'
print("before deleting black panther from the list:",heros)

heros.remove('black panther')
print("after deleting black panther from the list:",heros)

heros.insert(3,'black panther')
print("readding  black panther to the list:",heros)
```

```python
#4. Now you don't like thor and hulk because they get angry easily :)    So you want to remove thor and hulk from list and replace them with 
#doctor strange (because he is cool).   Do that with one line of code.

heros[1:3]=['doctor strange','doctor strange']
print(heros)
```

```python
#5. Sort the heros list in alphabetical order (Hint. Use dir() functions to list down all functions available in list)
#dir(heros)
heros.sort()
print(heros)
```

### 3. Create a list of all odd numbers between 1 and a max number. Max number is something you need to take from a user using input() function

```python
max_value=input("please add a number")
max_number=int(max_value)
list_number=[]
i=1
for i in range (max_number):
    if i % 2 !=0 :
        list_number.append(i)
    else:
        continue
print(list_number)
```
