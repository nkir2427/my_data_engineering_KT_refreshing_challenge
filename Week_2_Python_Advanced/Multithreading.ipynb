```python
import time
import threading
import math

def calc_square(numbers):
    print("calculate square numbers")
    for n in numbers:
        time.sleep(1)
        print('square:',n*n)

def calc_cube(numbers):
    print("calculate cube of numbers")
    for n in numbers:
        time.sleep(1)
        print('cube:',n*n*n)
        
def calc_squarert(numbers):
    print("calculate squareroot of numbers")
    for n in numbers:
        time.sleep(1)
        print('squarert:',math.sqrt(n))
```

```python
 

arr = [4,9,16,25]

t = time.time()

t1= threading.Thread(target=calc_square, args=(arr,))
t2= threading.Thread(target=calc_cube, args=(arr,))
t3= threading.Thread(target=calc_squarert, args=(arr,))

t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()

print("done in : ",time.time()-t)
print("Hah... I am done with all my work now!")
```
